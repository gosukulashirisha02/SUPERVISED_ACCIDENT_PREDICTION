```python
# Invalid speed limits
Accident_prediction.loc[
    ~Accident_prediction['Speed_Limit_kmh'].isin(
        [20, 30, 40, 50, 60, 70]
    ),
    'Speed_Limit_kmh'
] = np.nan

# Implausible temperatures
Accident_prediction.loc[
    (Accident_prediction['Temperature_C'] < -20) |
    (Accident_prediction['Temperature_C'] > 60),
    'Temperature_C'
] = np.nan

# Invalid visibility
Accident_prediction.loc[
    (Accident_prediction['Visibility_km'] < 0) |
    (Accident_prediction['Visibility_km'] > 20),
    'Visibility_km'
] = np.nan

# Invalid traffic volume
Accident_prediction.loc[
    (Accident_prediction['Traffic_Volume'] < 0) |
    (Accident_prediction['Traffic_Volume'] > 5000),
    'Traffic_Volume'
] = np.nan
```

```python
# Fill speed limits with the most frequent valid speed
Accident_prediction['Speed_Limit_kmh'] = (
    Accident_prediction['Speed_Limit_kmh'].fillna(
        Accident_prediction['Speed_Limit_kmh'].mode()[0]
    )
)

# Fill other numeric columns with their medians
columns_to_fill = [
    'Temperature_C',
    'Visibility_km',
    'Traffic_Volume'
]

for col in columns_to_fill:
    Accident_prediction[col] = (
        Accident_prediction[col].fillna(
            Accident_prediction[col].median()
        )
    )
```
